###**QUESTION 2**

```python
arr = np.array([1,2,3,4,5])
reversed_arr = np.array([],dtype = int)
for i in range(len(arr)-1,-1,-1):
  reversed_arr = np.append(reversed_arr,arr[i])

print(reversed_arr)
```

```python
def mostFreq(arr):
  unique_values = np.unique(arr)
  maxfreq = 0
  mostfreq = -1
  for value in unique_values:
    freq = np.sum(arr == value)
    if freq>maxfreq:
      maxfreq = freq
      mostfreq = value
  print("Most frequent no:", mostfreq)
  print("Frequence:",maxfreq)

x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
print("First array-",x)
mostFreq(x)
print("Second array-",y)
mostFreq(y)
```

```python
#Alternative:
def mostFreq(arr):
  unique_values,counts = np.unique(arr,return_counts = True)
  index = np.argmax(counts)
  mostfreq = unique_values[index]
  maxfreq = counts[index]
  print("Most frequent no:", mostfreq)
  print("Frequence:",maxfreq)
```

```python
ucs420 = np.array([ 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35])
print("Mean:", ucs420.mean())
print("Max:", ucs420.max())
print("Min:", ucs420.min())
print("Unique elements:", np.unique(ucs420))
```

```python
reshaped_ucs420 = ucs420.reshape(4,3)
print(reshaped_ucs420)
```

```python
resized_ucs420 = np.resize(ucs420,(2,3))
print(resized_ucs420)
```
