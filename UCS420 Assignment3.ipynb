```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)
```

```python
df.shape
```

```python
df.info()
```

```python
print(df.describe())
```

```python
df.head(3)
```

```python
df.tail(3)
```

```python
print("The Average Salary of employees is ", df['Salary'].mean())
print("The Total bonus paid to all employees is ", df['Bonus'].sum())
print("The Youngest employee's age is ", df['Age'].min())
print("The highest performance rating is ", df['Rating'].max())
```

```python
df.sort_values(by = 'Salary', ascending = False)
```

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance_Category"] = df["Rating"].apply(performance_category)
df
```

```python
df.isnull().sum()
```

```python
df.rename(columns={"Employee_ID": "ID"}, inplace=True)
df
```

```python
df[df["Years_of_Experience"] > 5]
```

```python
df["Tax"] = df["Salary"] * 0.10
df
```

```python
df.to_csv("modified_employees.csv", index=False)
```
