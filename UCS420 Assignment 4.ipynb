###**QUESTION 2**

```python
def calcScore(st):
  st = st.lower()
  query = st.split()
  scores = [0,0,0,0,0,0]
  n = len(df)
  for i in range(0,n):
    stored_keywords = df.loc[i,"keywords"].split()
    scores[i] = 0
    for word in query:
      if word in stored_keywords:
        scores[i] = scores[i] + 1
  return scores

ques = input("Enter your query: ")
FAQ_scores = calcScore(ques)
for i in range(0,6):
  print("Question ",i+1," :",FAQ_scores[i])
```

###**QUESTION 3**

```python
def same_category(category_name,df):
  questions = []
  for i in range(0,len(df)):
    if category_name == df.loc[i,"category"]:
      questions.append(df.loc[i,"question"])
  return questions

cat = input("Enter a category: ")
ques = same_category(cat,df)
print("Question of", cat, "category are:")
for i in range(0,len(ques)):
  print(ques[i])
```

###**QUESTION 6**

```python
def score_query(query, df):
    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())

        score = len(query_words & keywords) + len(query_words & question_words)

        if score > 0:
            results.append((index, score))

    # Sort by score in descending order
    results.sort(key=lambda x: x[1], reverse=True)

    if not results:
        print("No matching entries found.")
        return

    # Highest score
    highest_score = results[0][1]

    # Get ALL entries having the highest score
    top_matches = [index for index, score in results
                   if score == highest_score]

    print("Highest Score:", highest_score)
    print("\nMatching Entries:")

    for index in top_matches:
        print(df.loc[index])
        print()
```
